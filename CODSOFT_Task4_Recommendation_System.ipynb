```python

import pandas as pd

movies = {
    "Movie": [
        "Avengers",
        "Iron Man",
        "Thor",
        "The Notebook",
        "Titanic",
        "Interstellar",
        "Inception",
        "The Dark Knight",
        "Joker",
        "Avatar"
    ],
    "Genre": [
        "Action Sci-Fi",
        "Action Sci-Fi",
        "Action Fantasy",
        "Romance",
        "Romance Drama",
        "Sci-Fi Drama",
        "Sci-Fi Thriller",
        "Action Crime",
        "Crime Drama",
        "Sci-Fi Adventure"
    ]
}

df = pd.DataFrame(movies)

print("===================================")
print("   CODSOFT MOVIE RECOMMENDATION")
print("===================================")

print("\nMOVIE DATABASE")
display(df)

def recommend_movies(user_genre):
    recommendations = []

    for index, row in df.iterrows():
        if user_genre.lower() in row["Genre"].lower():
            recommendations.append(row["Movie"])

    if recommendations:
        print("\nRecommended Movies:")
        for movie in recommendations:
            print("-", movie)
    else:
        print("No movies found for this genre.")

print("\nExample Recommendations")

recommend_movies("Action")
recommend_movies("Romance")
recommend_movies("Sci-Fi")

print("\n===================================")
genre = input("Enter your favorite movie genre: ")
recommend_movies(genre)

print("\nRecommendation System is Ready!")
```
