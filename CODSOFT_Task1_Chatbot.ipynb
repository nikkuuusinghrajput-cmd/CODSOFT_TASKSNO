```python
import datetime
import random

print("================================")
print("       CODSOFT AI CHATBOT")
print("================================")

name = input("Bot: What is your name? ")
print("Bot: Nice to meet you,", name)
print("Bot: Type 'help' to see what I can do.")
print("Bot: Type 'bye' to exit.")

jokes = [
    "Why did the computer go to the doctor? Because it had a virus!",
    "Why do programmers prefer dark mode? Because light attracts bugs!",
    "Why was the computer cold? Because it left its Windows open!"
]

while True:
    user = input(name + ": ").lower().strip()

    if user in ["hello", "hi", "hey"]:
        print("Bot: Hello", name + "! How can I help you?")

    elif "your name" in user:
        print("Bot: I am CodSoft AI Chatbot.")

    elif "how are you" in user:
        print("Bot: I am doing great! How are you?")

    elif "what is ai" in user:
        print("Bot: AI stands for Artificial Intelligence.")

    elif "what is python" in user:
        print("Bot: Python is a popular programming language.")

    elif "time" in user or "date" in user:
        now = datetime.datetime.now()
        print("Bot: Current date and time:", now.strftime("%d-%m-%Y %I:%M %p"))

    elif "joke" in user:
        print("Bot:", random.choice(jokes))

    elif "thank" in user:
        print("Bot: You're welcome!")

    elif "help" in user:
        print("Bot: Ask me about AI, Python, time, date or jokes.")

    elif user in ["bye", "exit", "quit"]:
        print("Bot: Goodbye", name + "! Have a nice day!")
        break

    else:
        print("Bot: Sorry, I don't understand that.")
```
